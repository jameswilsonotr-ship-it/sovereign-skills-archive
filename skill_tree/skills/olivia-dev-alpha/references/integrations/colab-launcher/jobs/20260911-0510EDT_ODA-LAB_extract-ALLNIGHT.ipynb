````python

import io, json, os, sys, time, zipfile, tarfile, traceback, datetime
from pathlib import Path

PACKS = [["STORYBOARD_RECON_20260910.zip", "1dYjyx6S1Tp3pWvPC6hchxe6OV5gQ69DX", "zip"], ["LIV_PANE_ARCHIVE_20260910.zip", "1518UAB_yra8VEvfTiRef-fEnZrfjZ5pB", "zip"], ["plate-pack-normalized-20260910.zip", "1G1pHWZuIRVEt9XIanZ6jmjeQAXJ3G7yE", "zip"], ["EMIT-20260910.zip", "1ZokwsXIfogBZfL2sP5du4n_q2Pnn9M5H", "zip"], ["ANIME-COPPER-20260910.zip", "1b7xEj1Wft8DJQzWH4GzQ_4TICnZNQM0i", "zip"], ["GRID-BLONDE-20260910.zip", "1yVCuq80O-eMI6E9OYJ_iXZfzCL05K0tG", "zip"], ["cand-reel-133985.zip", "12NkSa8VC_Qe13eqisQjuZiq_F3pefn8Y", "zip"], ["IP-WQ-161-166.zip", "18LTJ0HrF7DMuNpB0W2JMzZ1vaj1Rjd4T", "zip"], ["IP-WQ-161-172.zip", "1t5Zal_pLKz5fwSANPnRjMy6YSbdMDl3W", "zip"], ["OLIVIA_ORCA_SESSION_20260901.zip", "1uvLj8Gs0hluOPVVUukLNUqiLZYr9ppqa", "zip"], ["SUNSET_20260910_FULLGO_textpack.zip", "1VDCA_HFcwM-akFAmpFqZASp9N4R_Sfxg", "zip"], ["SUNSET_20260910_EXTRACT.zip", "1tzJs4edsPQ2A2sZoVPv1my-XdR72YrfK", "zip"], ["SUNSET_20260910_artifacts_remainder.zip", "1WijYI_JPnWtdrW5hrrUvqm_1hRRMM9Xg", "zip"], ["SUNSET_20260910_artifacts_rendered.zip", "1qzgnWAPjrIJ9QvzZSChXETgP4XKihQmd", "zip"], ["LACEY-GROWN-20260910.zip", "1jwkbPpM3wlEi6PIz1XlZXqEsD3h5o_Je", "zip"], ["SUNSET_20260910_claim_runtime_visuals.zip", "1TbP1gLrJL-XIQ_RpzOxc8rESjYlEbUTf", "zip"], ["SUNSET_20260910_image_pipeline_visuals.zip", "1qMGCPsQvA2S0jQt6d_vARUr1GmB0Ecew", "zip"], ["SUNSET_20260910_pixels_images.zip", "1ookVe_nUn9ubLZlpoD5DDoqePZfGWzrX", "zip"], ["SUNSET_20260910_PIXELS.zip", "1VWXh1Jq3HMJ5vw8Se_Zo0L8fPJCR7S5M", "zip"], ["SUNSET_FULL_20260910_grok-five.tar.gz", "1q0yQh5Qoxp3Q-p2QmlpCCXe6dWQNzAO_", "tar.gz"], ["sunset_full_2026-09-10.zip", "14vU0nA6IwQGbzXXb7G6ZALBsnYkD0Fhg", "zip"], ["2026-09-10_SUNSET_TEXT_PACKAGES.zip", "1h68ZxmzfpGlxykGE1orYmZZyFm2Vxlms", "zip"], ["sunset_pixels_other_2026-09-10.zip", "1o2Ax6cuFTwu9d9XujuoUlBGtozH5s0ef", "zip"], ["sunset_pixels_imagine_rendered_2026-09-10.tar.gz", "1NVMHd5mUgXntJcTgC5TuAUC9XtBTlsTD", "tar.gz"], ["sunset_full_archive_2026-09-10.zip", "17Fhlp5gm0EzPrI7j6bnzHTzzGtVH5ZG0", "zip"], ["2026-09-10_SUNSET_PIXELS_rendered_and_attachments.zip", "1XlvwDgoNCsFGsPTQAidbBrlMs0Qlre6W", "zip"], ["sunset_pixels_2026-09-10.tar.gz", "19VZ0bGNYMUw5Wnh2GusMhn08I-fSg-QX", "tar.gz"], ["preview-only-message.tar.gz", "1fs7ape_GBlnpiZY3CiiXo7yVMbRSqghY", "tar.gz"], ["sunset-003-full-archive.tar.gz", "1e7yBIy4sH4zi10i2kh2_Ik1ikETk5PNH", "tar.gz"], ["SEQ-Crepax-pixel.tar.gz", "10yxkVArb7AwQ78WixYUNWUU6baPeW9IS", "tar.gz"], ["full-pixels-003-fat.tar.gz", "1OEd7nBVPkZPmdvy-UUE2of-2tMOTKvBh", "tar.gz"], ["imgpipe-memory-BIOS-004.tar.gz", "137uVQqiWHjcg9ezudMNWsUg2TvArz2bo", "tar.gz"]]
SPIN = "|/-\\"
ROOT = Path("/content/drive/MyDrive")
SHELF = ROOT / "12_ODA-LAB-NOTEBOOKS"
SHELF.mkdir(parents=True, exist_ok=True)
STAMP = time.strftime("%Y%m%d-%H%M")
DEST = SHELF / "extracts" / ("ALLNIGHT-" + STAMP)
DEST.mkdir(parents=True, exist_ok=True)
RUN = SHELF / "RUNNING.txt"
OKF = SHELF / "SUCCESS.txt"
FAILF = SHELF / "FAILED.txt"
LOG = DEST / "EXTRACT.LOG.md"
REC = DEST / "EXTRACT.RECEIPT.md"
IDX = DEST / "INDEX.md"
IDXJ = DEST / "INDEX.json"

def log(msg):
    line = time.strftime("%H:%M:%S") + " " + msg
    print(line, flush=True)
    with LOG.open("a") as f:
        f.write(line + "\n")

def heartbeat(state, extra=""):
    body = "state=%s\nstamp=%s\ndest=%s\n%s\n" % (
        state, time.strftime("%Y-%m-%d %H:%M:%S"), DEST, extra)
    RUN.write_text(body)
    (DEST / "RUNNING.txt").write_text(body)

def bar(pct, name, i, extra=""):
    pct = max(0, min(100, int(pct)))
    filled = pct // 5
    body = ("=" * filled + (">" if filled < 20 else ""))[:20].ljust(20)
    print("\r  [%s] %3d%% %s %s %s   " % (body, pct, SPIN[i % 4], name, extra), end="", flush=True)

print("+" + "-" * 62 + "+")
print("| ODA LAB ALLNIGHT   RUNNING                                 |")
print("| %s |" % str(DEST)[:60].ljust(60))
print("+" + "-" * 62 + "+")
heartbeat("RUNNING", "files=%d" % len(PACKS))
log("RUNNING dest=" + str(DEST))

idx_lines = ["# INDEX ALLNIGHT " + STAMP, "", "| # | name | id | kind |", "|---|---|---|---|"]
for i, (name, fid, kind) in enumerate(PACKS, 1):
    idx_lines.append("| %d | %s | `%s` | %s |" % (i, name, fid, kind))
IDX.write_text("\n".join(idx_lines) + "\n")
IDXJ.write_text(json.dumps([{"name":a,"id":b,"kind":c} for a,b,c in PACKS], indent=2))
log("INDEX written %d targets" % len(PACKS))

from google.colab import auth
auth.authenticate_user()
import google.auth
from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseDownload
creds, _ = google.auth.default()
svc = build("drive", "v3", credentials=creds)
log("drive v3 ok")

def drive_size(fid):
    meta = svc.files().get(fileId=fid, fields="id,name,size").execute()
    return int(meta.get("size") or 0), meta.get("name")

def pull(fid, dest_path, label):
    request = svc.files().get_media(fileId=fid)
    fh = io.FileIO(dest_path, "wb")
    dl = MediaIoBaseDownload(fh, request, chunksize=1024 * 1024)
    done = False
    i = 0
    while not done:
        status, done = dl.next_chunk()
        i += 1
        pct = int(status.progress() * 100) if status else 0
        bar(pct, label, i)
    fh.close()
    print()
    return Path(dest_path).stat().st_size

def extract_zip(src, out):
    out.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(src) as z:
        for info in z.infolist():
            z.extract(info, out)
            target = out / info.filename
            if target.is_file() and info.date_time[0] >= 1980:
                ts = datetime.datetime(*info.date_time).timestamp()
                os.utime(target, (ts, ts))
        return z.namelist()

def extract_tar(src, out):
    out.mkdir(parents=True, exist_ok=True)
    with tarfile.open(src, "r:*") as t:
        t.extractall(out)
        return t.getnames()

rows = []
ok = skip = fail = 0
t0_all = time.time()
for n, (name, fid, kind) in enumerate(PACKS, 1):
    print()
    t0 = time.time()
    heartbeat("RUNNING", "now=%s %d/%d" % (name, n, len(PACKS)))
    log("FILE %d/%d %s id=%s" % (n, len(PACKS), name, fid))
    out = DEST / Path(name).stem
    try:
        expected, dname = drive_size(fid)
    except Exception as e:
        log("META FAIL rc=2 " + repr(e))
        rows.append("- FAIL meta %s rc=2 %.1fs %s" % (name, time.time()-t0, e))
        fail += 1
        continue
    marker = out / "PACK.OK"
    local = Path("/content") / name
    if marker.exists() and out.exists():
        cnt = sum(1 for p in out.rglob("*") if p.is_file())
        log("SKIP finished files=%s expected=%s rc=0" % (cnt, expected))
        rows.append("- SKIP %s files=%s bytes=%s rc=0 %.1fs" % (name, cnt, expected, time.time()-t0))
        skip += 1
        continue
    if local.exists() and expected and local.stat().st_size == expected:
        sz = local.stat().st_size
        log("SKIP download same-size %s" % sz)
    else:
        try:
            sz = pull(fid, local, name)
        except Exception as e:
            log("DOWNLOAD FAIL rc=2 " + repr(e))
            traceback.print_exc()
            rows.append("- FAIL download %s rc=2 %.1fs %s" % (name, time.time()-t0, e))
            fail += 1
            continue
        if expected and sz != expected:
            log("SIZE MISMATCH got=%s expected=%s rc=5" % (sz, expected))
            rows.append("- FAIL size %s got=%s expected=%s rc=5" % (name, sz, expected))
            fail += 1
            continue
    try:
        names = extract_zip(local, out) if kind == "zip" else extract_tar(local, out)
        cnt = sum(1 for p in out.rglob("*") if p.is_file())
        dt = time.time() - t0
        marker.write_text("bytes=%s files=%s sec=%.1f\n" % (sz, cnt, dt))
        log("OK bytes=%s files=%s sec=%.1f rc=0" % (sz, cnt, dt))
        rows.append("- OK %s bytes=%s files=%s rc=0 %.1fs first=%s" % (name, sz, cnt, dt, names[:4]))
        ok += 1
    except Exception as e:
        log("EXTRACT FAIL rc=3 " + repr(e))
        traceback.print_exc()
        rows.append("- FAIL extract %s rc=3 %.1fs %s" % (name, time.time()-t0, e))
        fail += 1

elapsed = time.time() - t0_all
final = 0 if fail == 0 else 1
front = "---\nskill: olivia-dev-alpha\njob: ALLNIGHT\nenvelope: tui\nok: %d\nskip: %d\nfail: %d\nexit: %d\nsec: %.1f\ndest: %s\n---\n" % (ok, skip, fail, final, elapsed, DEST)
REC.write_text(front + "\n".join(rows) + "\n")
summary = "state=%s\nexit=%d\nok=%d skip=%d fail=%d\nsec=%.1f\nfinished=%s\ndest=%s\n" % (
    "SUCCESS" if final == 0 else "FAILED", final, ok, skip, fail, elapsed,
    time.strftime("%Y-%m-%d %H:%M:%S"), DEST)
if final == 0:
    OKF.write_text(summary)
    if FAILF.exists():
        FAILF.unlink()
else:
    FAILF.write_text(summary)
if RUN.exists():
    RUN.unlink()
r2 = DEST / "RUNNING.txt"
if r2.exists():
    r2.unlink()
(DEST / ("SUCCESS.txt" if final == 0 else "FAILED.txt")).write_text(summary)
log("DONE exit=%d ok=%d skip=%d fail=%d sec=%.1f" % (final, ok, skip, fail, elapsed))
print()
print(summary)
print("RECEIPT", REC)
print("EXIT", final)
````
